```python
# Handling ValueError
try:
    age = int("twenty")
    print("Age:", age)
except ValueError:
    print("Invalid value. Please enter a number.")
```

```python
# Handling TypeError
try:
    result = "10" + 5
    print("Result:", result)
except TypeError:
    print("Invalid operation. Different data types cannot be added.")
```

```python
# Basic try-except
try:
    number = 10
    divisor = 0
    result = number / divisor
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
# Try-except with user input
try:
    marks = int(input("Enter your marks: "))
    print("Marks:", marks)
except ValueError:
    print("Invalid input. Please enter a number.")
```

```python
# Try-except-else
try:
    number = 20
    divisor = 5
    result = number / divisor
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Result:", result)
```

```python
# Try-except-finally
try:
    number = 50
    divisor = 10
    result = number / divisor
    print("Result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero.")
finally:
    print("Program completed.")
```

```python
# Handling invalid data in a list
data = [10, 20, "abc", 40, 50]

for value in data:
    try:
        print("Value:", value + 5)
    except TypeError:
        print("Invalid data:", value)
```
